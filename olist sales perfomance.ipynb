```python
import pandas as pd

def convert_dataframe_numerics(df):
    """
    Attempts to convert each column in a DataFrame to numeric types (int/float).
    If a column contains non-numeric text, it is left untouched.
    """
    df_clean = df.copy()
    
    for col in df_clean.columns:
        # Try converting the entire column to numeric
        converted = pd.to_numeric(df_clean[col], errors='coerce')
        
        # If the column was originally non-null and conversion produced NO extra NaNs,
        # or if all non-null values were successfully converted, update the column.
        if converted.notna().sum() == df_clean[col].notna().sum() and df_clean[col].notna().sum() > 0:
            df_clean[col] = converted
            
    return df_clean

# --- APPLY TO ALL DATASETS IN GLOBAL MEMORY ---
for name, obj in list(globals().items()):
    if isinstance(obj, pd.DataFrame) and not name.startswith('_'):
        globals()[name] = convert_dataframe_numerics(obj)
        print(f"Updated column types for DataFrame: '{name}'")
```

```python
import unicodedata

def normalize_text_accents(text):
    """
    Strips Portuguese accents (e.g., 'São' -> 'Sao') 
    and standardizes casing and extra whitespaces.
    """
    if not isinstance(text, str):
        return text
    
    # Decompose unicode characters to separate letters from accents
    nfkd_form = unicodedata.normalize('NFKD', text)
    # Rebuild string keeping only base ASCII letters
    clean_str = "".join([c for c in nfkd_form if not unicodedata.combining(c)])
    
    # Standardize spaces and apply title case
    return clean_str.strip().title()

# Map of city columns to clean across your loaded DataFrames
city_columns_to_fix = [
    ('customers', 'customer_city'),
    ('sellers', 'seller_city'),
    ('geolocation', 'geolocation_city')
]

# Apply accent normalization
for df_name, col in city_columns_to_fix:
    if df_name in globals():
        globals()[df_name][col] = globals()[df_name][col].apply(normalize_text_accents)
        print(f"Normalized city names in '{df_name}.{col}'")
```

```python
import numpy as np
import pandas as pd

# 1. Map columns that MUST be strictly non-negative (>= 0)
non_negative_schema = {
    'order_items': ['price', 'freight_value', 'order_item_id'],
    'order_payments': ['payment_value', 'payment_sequential', 'payment_installments'],
    'order_reviews': ['review_score'],
    'products': [
        'product_photos_qty', 'product_name_lenght', 'product_description_lenght',
        'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm'
    ]
}

# 2. Define columns where negative numbers ARE allowed (geographical coordinates in Brazil)
valid_negative_schema = {
    'geolocation': ['geolocation_lat', 'geolocation_lng']
}

print("=== DETECTION & CLEANING OF INVALID NEGATIVE VALUES ===\n")

total_negatives_cleaned = 0

# Scan non-negative target columns
for df_name, columns in non_negative_schema.items():
    if df_name in globals() and isinstance(globals()[df_name], pd.DataFrame):
        df = globals()[df_name]
        
        for col in columns:
            if col in df.columns:
                # Detect negative numbers (< 0)
                neg_mask = df[col] < 0
                neg_count = neg_mask.sum()
                
                if neg_count > 0:
                    total_negatives_cleaned += neg_count
                    print(f"⚠️ DETECTED: {neg_count} invalid negative value(s) in '{df_name}.{col}'")
                    # Replace invalid negatives with NaN
                    df.loc[neg_mask, col] = np.nan
                    print(f"   ↳ Cleaned: Replaced negative values in '{df_name}.{col}' with NaN.")
                else:
                    print(f"✅ CLEAN: '{df_name}.{col}' has 0 negative values.")

print("\n" + "="*60)
print("=== PRESERVED COLUMNS (WHERE NEGATIVES ARE VALID) ===")

# Verify coordinate columns stay untouched
for df_name, columns in valid_negative_schema.items():
    if df_name in globals() and isinstance(globals()[df_name], pd.DataFrame):
        df = globals()[df_name]
        for col in columns:
            if col in df.columns:
                neg_count = (df[col] < 0).sum()
                print(f"📍 '{df_name}.{col}' contains {neg_count:,} valid negative coordinates (LEFT UNTOUCHED).")

print("="*60)
if total_negatives_cleaned == 0:
    print("🎉 RESULT: No invalid negative values were found in your data!")
else:
    print(f"🎉 RESULT: Successfully removed {total_negatives_cleaned} invalid negative value(s).")
```

```python
import numpy as np
import pandas as pd

# 1. Filter for negative reviews using master_df
negative_reviews = master_df[master_df['review_score'] <= 2].copy()

# 2. Function to categorize complaints based on Portuguese text keywords
def categorize_complaint(text):
    if pd.isna(text):
        return 'No Comment'
    text = str(text).lower()
    if any(k in text for k in ['demor', 'atras', 'nao chegou', 'prazo']):
        return 'Late / Delayed'
    elif any(k in text for k in ['quebrad', 'estragad', 'defeito', 'danificad']):
        return 'Damaged / Defective'
    elif any(k in text for k in ['errado', 'diferente', 'faltou', 'veio']):
        return 'Wrong / Incomplete'
    elif any(k in text for k in ['nao recebi', 'nunca']):
        return 'Not Delivered'
    else:
        return 'Other Complaint'

# Combine title and message for keyword matching
comment_text = (
    negative_reviews['review_comment_title'].fillna('') + " " + 
    negative_reviews['review_comment_message'].fillna('')
)

# 3. Apply categorization
negative_reviews['Complaint_Category'] = comment_text.apply(categorize_complaint)

# 4. Deduplicate by review_id before counting (prevents multi-item order inflation)
complaint_counts = (
    negative_reviews.drop_duplicates(subset=['review_id'])['Complaint_Category']
    .value_counts()
    .reset_index(name='Negative Review Count')
)

print(complaint_counts)
```
